```python
# 1. Crea una lista llamada numeros_pares, que contenga los números del 1 al 100 y que contenga saltos de 2 en 2
numeros_pares=list(range(2,101,2))
print(numeros_pares)
```

```python
# 2.Agrega los números 102, 104 y 106 al final de la lista numeros_pares y luego elimina los primeros 5 números de la lista numeros_pares. Muestra la lista resultante.
# Agregar los valores 102,104 y 106 a numeros_pares
numeros_pares.extend([102,104,106])
# Eliminar los primeros 5 datos
del numeros_pares[:5]
# Resultado
print(numeros_pares)
```

```python
#Crea una lista llamada resultante que contenga la raíz cuadrada de cada elemento de la lista numeros_pares usando list comprehension.
import math
#Ya se tiene la lista numeros_pares
#Crear lista resultante usando list comprhensión
resultante = [math.sqrt(num) for num in numeros_pares]
#Resultado
print(resultante)
```

```python
#Dada la siguiente tupla tupla_x = (“alumno”,”Pedro”,15), desempaqueta la tupla tupla_x en tres variables: x, y, z, e imprime la siguiente estructura “El alumno Pedro obtuvo la calificación de 15 usando las variables x, y, z”.
tupla_x=("alumno","pedro",15)
#Desempaquetar Tupla_x
x,y,z = tupla_x
#Resultado
print(f"El {x} {y} obtuvo la calificación de {z}")
```

```python
#Dado el siguiente diccionario notas = {“Programacion”:15.9, “Matematica”: 15.1, “Fisica”:15.6}, escribe un bucle for que imprima cada clave y su valor en el diccionario con la siguiente estructura “Mis calificaciones en el curso de Programacion es 15.9”

notas = {"Programacion":15.9,"matematica":15.1,"Física":15.6}
#Bucle for para iterar sobre el diccionario creado
for curso, calificación in notas.items():
#Resultado
   print(f"Mis calificaciones en el curso de {curso} es {calificación}")
```

```python
#Crea una matriz 1D que contenga los primeros 16 elementos de la lista numeros_pares (creada en la pregunta 1), modifique las dimensiones y convertiela en una matriz 2D con shape(4,4), calcula la suma total de los elementos de la matriz matriz 2D.
import numpy as np
#Crear la lista de los primeros 16 números pares
numeros_pares_2 = list(range(2,101,2))

primeros_16_elementos = numeros_pares_2[:16]
#Convertir lista en una matriz 2D utilizando shape (4,4)
matriz_2D = np.array(primeros_16_elementos).reshape(4,4)
#Calcular Suma total de elementos de la matriz 2D
total = np.sum(matriz_2D)

#Resultados
print("Matriz 2D:")
print(matriz_2D)
print("\nSuma total de elementos:", total)
```
